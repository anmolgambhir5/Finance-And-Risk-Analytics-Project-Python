```python
stock_means = stock_returns.mean(axis = 0)
stock_means
```

```python
stock_sd = stock_returns.std(axis = 0)
stock_sd
```

```python
df2 = pd.DataFrame({'Average':stock_means, 'Volatility': stock_sd})
df2
```

```python
#plt.scatter(stock_sd, stock_means, edgecolors='r')
x=df2['Volatility']
y=df2['Average']
plot = plt.scatter(x,y)
labels = ['Infy', 'IH', 'M&M', 'Axis', 'SAIL','Shree','Sun','JS','Idea','Jet']
# Label each point
for i, label in enumerate(labels):
    plt.annotate(label, (x[i], y[i]))
plt.xlabel('Volatility')
plt.ylabel('Average')
plt.show()
```
